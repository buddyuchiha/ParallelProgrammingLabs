Разница на матрицах с большим размером между суперкомпьютером "Сергей Королёв" с технологией MPI и обычного компьютера с тенологией MPI

```python
mpi_korolev_time_path = 'ParallelProgrammingLab1\\output\\mpi_korolev_time\\8.txt'
path = 'ParallelProgrammingLab1\\output\\mpi_time\\mpi_time_8\\'
mpi_time = list()
compare_mpi = list()

for size in range (100, 1100, 100):
    time_path = path + f'mpi_time{size}.txt'
    num = read_time(time_path)
    mpi_time.append(num)

temp_mpi_korolev_time = read_mpi_time(mpi_korolev_time_path)
mpi_korolev_time = [int(num) for num in temp_mpi_korolev_time.split()]

for j in range(len(mpi_time)):
    compare_mpi.append(mpi_time[j] // mpi_korolev_time[j]) 

x = [int(num) for num in range (100, 1100, 100)]
y = compare_mpi

plt.plot(x, y, label="result", color="blue")
plt.xlabel("Размер матрицы")
plt.ylabel("Разница")
plt.title(f"Во сколько раз суперкомпьютер Сергей Королёв быстрее ")
```

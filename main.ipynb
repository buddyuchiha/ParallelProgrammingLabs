__Вывод__

Можем заметить, что количество ядер существенно влияет на время выполнение задач.

Также стоит отметить, что все изменения происходят у матриц с большим размером, когда у маленьких размеров все происходит непредсказуемо.

Самую большую соседнюю разницу мы получили, когда использовал вместо двух ядер - четыре ядра и эта разница составила 1.75

```python
small_matrix = dict()
high_matrix = dict()

for core in range (2, 10, 2):
    small_time = read_time(f'ParallelProgrammingLab1\\output\\open_mp\\open_mp_{core}\\time100.txt')
    high_time = read_time(f'ParallelProgrammingLab1\\output\\open_mp\\open_mp_{core}\\time1000.txt')
    small_matrix[core] = small_time 
    high_matrix[core] = high_time

small_pair = min(small_matrix.items(), key=lambda x: x[1])
high_pair = min(high_matrix.items(), key=lambda x: x[1])

print(f"Самое оптимизированное количество ядер для матриц размером 100 = {small_pair[0]}, со временем = {small_pair[1]}")
print(f"Самое оптимизированное количество ядер для матриц размером 1000 = {high_pair[0]}, со временем = {high_pair[1]}")
```
